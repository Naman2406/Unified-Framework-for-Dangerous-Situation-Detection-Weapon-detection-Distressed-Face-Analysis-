```python
from ultralytics import YOLO
import cv2
import math 
from tensorflow.keras.models import load_model
import numpy as np
from matplotlib import pyplot as plt
emot_classes = ["distressed", "undistressed"]

# start webcam
cap = cv2.VideoCapture(0)
# cap.set(3, 640)
# cap.set(4, 480)

# model
obj_model = YOLO('C:/Users/ruchi/OneDrive/Desktop/Capstone/Code/Combined Result/best.pt')
face_model=load_model('C:/Users/ruchi/OneDrive/Desktop/Capstone/Code/Combined Result/model.h5')

# object classes
dang_classes = ['alcohol', 'gesture', 'blood', 'cigarette', 'gun', 'knife']
face_cascade = cv2.CascadeClassifier(cv2.data.haarcascades + 'haarcascade_frontalface_default.xml')


while True:
    success, img = cap.read()

    gray_img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

    # Detect faces in the image
    faces = face_cascade.detectMultiScale(gray_img, scaleFactor=1.1, minNeighbors=5)

    # Loop over each detected face
    # img2=cv2.imread(img_path)
    for (x, y, w, h) in faces:
        # Extract the face region
        face_roi = img[y:y+h, x:x+w]

        # Resize the face to match the expected input size of your emotion model
        resized_face = cv2.resize(face_roi, (256, 256))

        # Expand dimensions to match the model's input shape (add the batch dimension)
        input_data = np.expand_dims(resized_face, axis=0)

        # Normalize pixel values to the range [0, 1]
        input_data = input_data / 255.0

        # Predict emotion for the face
        emotion_prediction = face_model.predict(input_data)

        # Assuming emotion_prediction contains your emotion class probabilities
        predicted_emotion = emot_classes[np.argmax(emotion_prediction)]

        # img2 = cv2.cvtColor(img2, cv2.COLOR_BGR2RGB)  # Convert BGR to RGB


        # Draw rectangles and text on the image
        # img=cv2.cvtColor(img, cv2.RGB2BGR)
        cv2.rectangle(img, (x, y), (x+w, y+h), (255, 0, 0), 2)
        cv2.putText(img, f'Emotion: {predicted_emotion}', (x, y-10), cv2.FONT_HERSHEY_SIMPLEX, 0.4, (255, 0, 0), 2)

    # coordinates
    results = obj_model(img, stream=True)
    for r in results:
        boxes = r.boxes

        for box in boxes:
            # bounding box
            x1, y1, x2, y2 = box.xyxy[0]
            x1, y1, x2, y2 = int(x1), int(y1), int(x2), int(y2) # convert to int values

            # put box in cam
            cv2.rectangle(img, (x1, y1), (x2, y2), (0, 255, 0), 3)

            # confidence
            confidence = math.ceil((box.conf[0]*100))/100
            print("Confidence --->",confidence)

            # class name
            cls = int(box.cls[0])
            print("Class name -->", dang_classes[cls])

            # object details
            org = [x1, y1-10]
            font = cv2.FONT_HERSHEY_SIMPLEX
            fontScale = 1
            color = (0, 255, 0)
            thickness = 2

            cv2.putText(img, dang_classes[cls], org, font, fontScale, color, thickness)

            
    cv2.namedWindow("Webcam", cv2.WINDOW_NORMAL)
    cv2.resizeWindow("Webcam", 800, 800) 
    cv2.imshow('Webcam', img)
    if cv2.waitKey(1) == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```
